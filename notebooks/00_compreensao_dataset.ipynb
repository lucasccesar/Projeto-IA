## 3. Relação entre `idLobbyGame` e `idRoom`

A documentação identifica `idLobbyGame` como o identificador da partida e `idRoom` como o identificador da sala.

A análise identificou a seguinte distribuição:

| Tipo de partida | Quantidade |
|---|---:|
| Partidas com uma room | 167.322 |
| Partidas com duas rooms | 5.589 |
| Partidas com três ou mais rooms | 0 |
| **Total de partidas** | **172.911** |

**Principais descobertas:**

- Cada `idRoom` está associado a um único `idLobbyGame` nos dados disponíveis.
- Uma partida pode estar associada a uma ou duas rooms.
- Todos os registros de uma mesma partida compartilham o mapa e o horário.
- Uma mesma room pode conter registros de jogadores vencedores e perdedores. Portanto, `idRoom` não identifica exclusivamente uma equipe.

### Jogadores registrados por room

| Jogadores distintos | Quantidade de rooms |
|---|---:|
| 1 | 172.966 |
| 2 | 5.416 |
| 3 | 118 |
| **Total** | **178.500** |

Não existem rooms com seis ou dez jogadores registrados no dataset.

Isso não significa que as partidas reais possuíam somente esses participantes. O conjunto de dados pode conter apenas uma parte dos jogadores que disputaram cada confronto.

**Conclusão:** `idLobbyGame` será utilizado como identificador da partida. `idRoom` será preservado como identificador da sala, mas sua função operacional específica na plataforma ainda precisa ser confirmada.

# 2 - Valores nulos

 Provável Causa Raiz

* **Origem:** Instabilidade/falha técnica no plugin de telemetria do servidor durante partidas específicas (`idLobbyGame`).
* **Evidência:** Jogadores que possuem eliminações registradas (`qtKill` > 0 e `qtHs` > 0) aparecem com acertos em membros corporais como `NaN`, confirmando que a ação ocorreu na partida, mas o log do servidor não foi enviado para a base.
