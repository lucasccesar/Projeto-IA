## 1. Visão Geral do Dataset
* **Total de Linhas (Registros):** 184.152
* **Total de Colunas:** 38
* **Linhas Sem Nulos (100% Completas):** 183.447
* **Linhas com Pelo Menos 1 Nulo:** 705 

## 2. Mapeamento de Valores Ausentes por Coluna

Das 38 colunas do dataset, **24 colunas possuem dados 100% preenchidos** e **14 colunas contêm valores ausentes (`NaN`)**, divididas em dois blocos principais:

### 🔹 Bloco 1: Estatísticas de Acertos Anatômicos e Ações Especiais (705 Nulos)
Contêm **705 valores ausentes**:
* `qtSurvived`
* `qtTrade`
* `qtFlashAssist`
* `qtHitHeadshot`
* `qtHitChest`
* `qtHitStomach`
* `qtHitLeftArm`
* `qtHitRightArm`
* `qtHitLeftLeg`
* `qtHitRightLeg`

### 🔹 Bloco 2: Estatísticas Detalhadas de Dano e Movimentação (120 Nulos)
Contêm **120 valores ausentes**:
* `qtTk`
* `qtTkAssist`
* `qtHits`
* `qtLastAlive`


## 3. Relação e Estrutura Entre os Nulos (Aninhamento)

* **Aninhamento Total:** Os nulos do **Bloco 2 (120 linhas)** estão **100% contidos** dentro do **Bloco 1 (705 linhas)**.
* **Detalhamento dos Grupos:**
  * **585 linhas:** Têm registros básicos (`qtHits`, `qtTk`), mas falharam em registrar acertos corporais e suporte (`qtHitChest`, `qtSurvived`).
  * **120 linhas:** Apresentam falha total nos dois blocos de métricas ao mesmo tempo.
  * **0 linhas:** Têm nulos isolados fora dessas 705 partidas.


## 4. Causa Raiz Identificada

* **Origem:** Instabilidade/falha técnica no plugin de telemetria do servidor durante partidas específicas (`idLobbyGame`).
* **Evidência:** Jogadores que possuem eliminações registradas (`qtKill` > 0 e `qtHs` > 0) aparecem com acertos em membros corporais como `NaN`, confirmando que a ação ocorreu na partida, mas o log anatômico/avançado do servidor não foi gravado/enviado para a base.
